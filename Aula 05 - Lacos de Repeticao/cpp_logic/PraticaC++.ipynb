```python
#While

#include <iostream>

int main() {
    std::cout << "\nAdivinhe o número com 'while' em C++!" << std::endl;
    int numero_secreto = 7;
    int numero_digitado = 0;

    while (numero_digitado != numero_secreto) {
        std::cout << "Digite um número: ";
        std::cin >> numero_digitado;
        if (numero_digitado != numero_secreto) {
            std::cout << "Número incorreto. Tente novamente." << std::endl;
        }
    }

    std::cout << "Parabéns! Você acertou!" << std::endl;
    return 0;
}
```

```python
#Do-while

#include <iostream>
using namespace std;

int main() {
    int estoque;
    do {
        cout << "Digite a quantidade (não pode ser negativa): ";
        cin >> estoque;
    } while (estoque < 0);

    cout << "Você digitou o número correto? ";
    return 0;
}
```

```python
#For

#include <iostream>

int main() {
    std::cout << "Contando com 'for' em C++:" << std::endl;
    for (int i = 1; i <= 5; i++) {
        std::cout << i << std::endl;
    }
    return 0;
}
```

```python
#Lista com For

#include <iostream>
#include <vector>
#include <string>
using namespace std;

int main() {
    vector<string> nomes = {"Ana", "Bruno", "Carla"};
    cout << "\nIterando sobre um vetor em C++:" << endl;
    for (const string& nome : nomes) {
        cout << nome << endl;
    }
    return 0;
}
```
