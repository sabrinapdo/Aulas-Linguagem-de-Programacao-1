```python
#Desafio 1

#include <iostream>
#include <vector>
using namespace std;

int main() {

    vector<string> nomes = {"Shakespeare", "Jane", "Machado", "Kafka", "Fernando"};

    for (int i = 1; i <= 10; i++) {
        cout << i << endl;

        if (i <= 5) {
            cout << nomes[i - 1] << endl;
        }
    }

    return 0;
}
```

```python
#Desafio 2

#include <iostream>
using namespace std;

int main() {
    int numero = 0;

    while (numero <= 10) {
        cout << "Digite um numero: ";
        cin >> numero;
    }

    cout << "Obrigado! Voce digitou um numero maior que 10." << endl;

    return 0;
}
```

```python
#Desafio 3

#include <iostream>
using namespace std;

int main() {
    float nota, soma = 0;
    
    for(int i = 1; i <= 5; i++) {
        cout << "Digite a nota " << i << ": ";
        cin >> nota;
        soma += nota;
    }

    float media = soma / 5;
    cout << "Media final: " << media << endl;

    return 0;
}
```
