```python
#Desafio 2

numero = 0

while numero <= 10:
    numero = int(input("Digite um numero: "))

print("Obrigado! Você digitou um número maior que 10.")
```

```python
#Desafio 3

soma = 0

for i in range(1, 6):
    nota = float(input(f"Digite a nota {i}: "))
    soma += nota

media = soma / 5
print("Média final:", media)
```

```python
#Desafio 4

numero_secreto = 7
palpite = 0

while palpite != numero_secreto:
    palpite = int(input("Adivinhe o número: "))

    if palpite > numero_secreto:
        print("Tente um número menor!")
    elif palpite < numero_secreto:
        print("Tente um número maior!")

print("Parabéns! Você acertou!")
```

```python
#Desafio 5

senha_correta = "senha123"
senha_digitada = ""

while senha_digitada != senha_correta:
    senha_digitada = input("Digite a senha: ")

    if senha_digitada != senha_correta:
        print("Senha incorreta!")

print("Acesso liberado! Bem-vindo!")
```

```python
#Desafio 6

<?php
$produtos = [];

$opcao = 0;

while ($opcao != 3) {
    echo "\n--- MENU ---\n";
    echo "1 - Adicionar produto\n";
    echo "2 - Listar produtos\n";
    echo "3 - Sair\n";
    echo "Escolha: ";
    $opcao = (int) fgets(STDIN);

    if ($opcao == 1) {
        echo "Nome do produto: ";
        $nome = trim(fgets(STDIN));

        do {
            echo "Quantidade: ";
            $quantidade = (int) fgets(STDIN);
            if ($quantidade < 0) {
                echo "Quantidade não pode ser negativa!\n";
            }
        } while ($quantidade < 0);

        echo "Preço: ";
        $preco = (float) fgets(STDIN);

        $produtos[] = [
            "nome" => $nome,
            "quantidade" => $quantidade,
            "preco" => $preco
        ];

        echo "Produto adicionado!\n";

    } elseif ($opcao == 2) {
        if (empty($produtos)) {
            echo "Estoque vazio!\n";
        } else {
            echo "\n--- ESTOQUE ---\n";
            foreach ($produtos as $p) {
                echo "Nome: {$p['nome']} | Qtd: {$p['quantidade']} | Preço: {$p['preco']}\n";
            }
        }

    } elseif ($opcao == 3) {
        echo "Saindo...\n";
    } else {
        echo "Opção inválida!\n";
    }
}
?>
```
