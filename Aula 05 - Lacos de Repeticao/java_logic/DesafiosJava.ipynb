```python
#Desafio 1

public class Desafio1 {
    public static void main(String[] args) {

        String[] nomes = {"Shakespeare", "Jane", "Machado", "Kafka", "Fernando"};

        for (int i = 1; i <= 10; i++) {
            System.out.println(i);

            if (i <= 5) {
                System.out.println(nomes[i - 1]);
            }
        }
    }
}
```

```python
#Desafio 2

import java.util.Scanner;

public class Desafio2 {
    public static void main(String[] args) {
        Scanner scanner = new Scanner(System.in);
        int numero = 0;

        while (numero <= 10) {
            System.out.print("Digite um numero: ");
            numero = scanner.nextInt();
        }

        System.out.println("Obrigado! Voce digitou um numero maior que 10.");

        scanner.close();
    }
}
```

```python
#Desafio 3

import java.util.Scanner;

public class Desafio3 {
    public static void main(String[] args) {
        Scanner scanner = new Scanner(System.in);
        double nota, soma = 0;

        for(int i = 1; i <= 5; i++) {
            System.out.print("Digite a nota " + i + ": ");
            nota = scanner.nextDouble();
            soma += nota;
        }

        double media = soma / 5;
        System.out.println("Media final: " + media);

        scanner.close();
    }
}
```

```python
#Desafio 4

import java.util.Scanner;

public class Desafio4 {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        int numeroSecreto = 7;
        int palpite = 0;

        while (palpite != numeroSecreto) {
            System.out.print("Adivinhe o número: ");
            palpite = sc.nextInt();

            if (palpite > numeroSecreto) {
                System.out.println("Tente um número menor!");
            } else if (palpite < numeroSecreto) {
                System.out.println("Tente um número maior!");
            }
        }

        System.out.println("Parabéns! Você acertou!");
        sc.close();
    }
}
```

```python
#Desafio 5

import java.util.Scanner;

public class Desafio5 {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        String senhaCorreta = "senha123";
        String senhaDigitada = "";

        while (!senhaDigitada.equals(senhaCorreta)) {
            System.out.print("Digite a senha: ");
            senhaDigitada = sc.nextLine();

            if (!senhaDigitada.equals(senhaCorreta)) {
                System.out.println("Senha incorreta!");
            }
        }

        System.out.println("Acesso liberado! Bem-vindo!");
        sc.close();
    }
}
```

```python
#Desafio 6

produtos = []

while True:
    print("\n--- MENU ---")
    print("1 - Adicionar produto")
    print("2 - Listar produtos")
    print("3 - Sair")

    opcao = input("Escolha: ")

    if opcao == "1":
        nome = input("Nome do produto: ")

        while True:
            quantidade = int(input("Quantidade: "))
            if quantidade >= 0:
                break
            print("Quantidade não pode ser negativa!")

        preco = float(input("Preço: "))

        produtos.append({
            "nome": nome,
            "quantidade": quantidade,
            "preco": preco
        })

        print("Produto adicionado!")

    elif opcao == "2":
        if not produtos:
            print("Estoque vazio!")
        else:
            print("\n--- ESTOQUE ---")
            for p in produtos:
                print(f"Nome: {p['nome']} | Qtd: {p['quantidade']} | Preço: {p['preco']}")

    elif opcao == "3":
        print("Saindo...")
        break

    else:
        print("Opção inválida!")
```
