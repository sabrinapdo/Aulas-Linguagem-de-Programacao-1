```python
#While

import java.util.Scanner;

class While {
    public static void main(String[] args) {
        System.out.println("\nAdivinhe o número com 'while' em Java!");
        Scanner scanner = new Scanner(System.in);
        int numero_secreto = 7;
        int numero_digitado = 0;

        while (numero_digitado != numero_secreto) {
            System.out.print("Digite um número: ");
            numero_digitado = scanner.nextInt();
            if (numero_digitado != numero_secreto) {
                System.out.println("Número incorreto. Tente novamente.");
            }
        }

        System.out.println("Parabéns! Você acertou!");
        scanner.close();
    }
}
```

```python
#Do-while

import java.util.Scanner;

class DoWhile{
    public static void main(String[] args) {
        Scanner scanner = new Scanner(System.in);
        int estoque;

        do {
            System.out.print("Digite a quantidade (não pode ser negativa): ");
            estoque = scanner.nextInt();
        } while (estoque < 0);

        System.out.println("Parabéns, você digitou o número errado. ");
        scanner.close();
    }
}
```

```python
#For

class For {
    public static void main(String[] args) {
        System.out.println("Contando com 'for' em Java:");
        for (int i = 1; i <= 5; i++) {
            System.out.println(i);
        }
    }
}
```

```python
#Lista com For

import java.util.ArrayList;

class ListaFor {
    public static void main(String[] args) {
        ArrayList<String> nomes = new ArrayList<>();
        nomes.add("Ana");
        nomes.add("Bruno");
        nomes.add("Carla");

        System.out.println("\nIterando sobre uma lista em Java:");
        for (String nome : nomes) {
            System.out.println(nome);
        }
    }
}
```
