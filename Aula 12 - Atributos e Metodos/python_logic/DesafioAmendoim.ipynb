```python
class Insumo:

    def __init__(self, nome, quantidade, mesValidade):

        self.__nome = nome
        self.__quantidade = quantidade

        if mesValidade >= 1 and mesValidade <= 12:
            self.__mesValidade = mesValidade
        else:
            print("ALERTA: Mês inválido! Definindo validade padrão.")
            self.__mesValidade = 1

    def setMesValidade(self, mes):

        if mes >= 1 and mes <= 12:
            self.__mesValidade = mes
            print("Mês de validade atualizado com sucesso.")
        else:
            print("ALERTA: Valor inválido! Operação bloqueada.")

    def estaValido(self, mesAtual):

        if self.__mesValidade < mesAtual:
            print("ALERTA: Risco de Amendoim Murcho! Produto Vencido")
            return False

        return True

    def mostrarDados(self):

        print("\n--- DADOS DO INSUMO ---")
        print("Nome:", self.__nome)
        print("Quantidade:", self.__quantidade)
        print("Mês de validade:", self.__mesValidade)


insumo1 = Insumo("Amendoim", 50, 8)

insumo1.mostrarDados()

insumo1.setMesValidade(10)

insumo1.setMesValidade(15)

print("\nProduto válido?", insumo1.estaValido(7))

print("\nProduto válido?", insumo1.estaValido(11))
```
