```python
#Classe: Livro

#Atributos: título, autor e gênero.

#Métodos: abrir, fechar, emprestar e devolver.
```

```python
class Livro:

    def __init__(self, titulo, autor, genero):
        self.titulo = titulo
        self.autor = autor
        self.genero = genero

    def abrir(self):
        print("Livro aberto.")

    def fechar(self):
        print("Livro fechado.")

    def emprestar(self):
        print("Livro emprestado.")

    def devolver(self):
        print("Livro devolvido.")


livro1 = Livro("A Abadia de Northanger", "Jane Austen", "Romance")
livro2 = Livro("Persuasao", "Jane Austen", "Romance")
livro3 = Livro("Emma", "Jane Austen", "Romance")

livro1.abrir()
livro1.fechar()
livro1.emprestar()
livro1.devolver()

livro2.abrir()
livro2.fechar()
livro2.emprestar()
livro2.devolver()

livro3.abrir()
livro3.fechar()
livro3.emprestar()
livro3.devolver()
```
