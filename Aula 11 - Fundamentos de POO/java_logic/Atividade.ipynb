```python
class Livro {

    String titulo;
    String autor;
    String genero;

    void abrir() {
        System.out.println("Livro aberto.");
    }

    void fechar() {
        System.out.println("Livro fechado.");
    }

    void emprestar() {
        System.out.println("Livro emprestado.");
    }

    void devolver() {
        System.out.println("Livro devolvido.");
    }
}

public class Main {

    public static void main(String[] args) {

        Livro livro1 = new Livro();
        livro1.titulo = "A Abadia de Northanger";
        livro1.autor = "Jane Austen";
        livro1.genero = "Romance";

        Livro livro2 = new Livro();
        livro2.titulo = "Persuasao";
        livro2.autor = "Jane Austen";
        livro2.genero = "Romance";

        Livro livro3 = new Livro();
        livro3.titulo = "Emma";
        livro3.autor = "Jane Austen";
        livro3.genero = "Romance";

        livro1.abrir();
        livro2.emprestar();
        livro3.devolver();
    }
}
```
