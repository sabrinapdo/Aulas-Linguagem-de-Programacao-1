```python
#include <iostream>
#include <string>

using namespace std;

class Livro {

public:
    string titulo;
    string autor;
    string genero;

    void abrir() {
        cout << "Livro aberto." << endl;
    }

    void fechar() {
        cout << "Livro fechado." << endl;
    }

    void emprestar() {
        cout << "Livro emprestado." << endl;
    }

    void devolver() {
        cout << "Livro devolvido." << endl;
    }
};

int main() {

    Livro livro1;
    livro1.titulo = "A Abadia de Northanger";
    livro1.autor = "Jane Austen";
    livro1.genero = "Romance";

    Livro livro2;
    livro2.titulo = "Persuasao";
    livro2.autor = "Jane Austen";
    livro2.genero = "Romance";

    Livro livro3;
    livro3.titulo = "Emma";
    livro3.autor = "Jane Austen";
    livro3.genero = "Romance";

    livro1.abrir();
    livro1.fechar();
    livro1.emprestar();
    livro1.devolver();

    livro2.abrir();
    livro2.fechar();
    livro2.emprestar();
    livro2.devolver();

    livro3.abrir();
    livro3.fechar();
    livro3.emprestar();
    livro3.devolver();

    return 0;
}
```
