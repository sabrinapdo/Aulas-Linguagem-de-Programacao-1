```python
# Função 1
def verificar_estoque_critico(quantidade):
    if quantidade < 5:
        return " [REPOSIÇÃO NECESSÁRIA]"
    return ""

# Função 2
def exibir_cabecalho():
    print("\n   SORVETERIA   ")
    print("---------------------------")

nomes = []
quantidades = []

i = 1

# while no cadastro
while i <= 3:
    nome = input(f"Digite o nome do {i}º produto: ")
    qtd = int(input(f"Digite a quantidade de {nome}: "))

    nomes.append(nome)
    quantidades.append(qtd)

    i += 1

exibir_cabecalho()

# while no relatório
j = 0
while j < len(nomes):
    alerta = verificar_estoque_critico(quantidades[j])

    print(f"Produto: {nomes[j]}{alerta} | Estoque: {quantidades[j]} unidades.")

    j += 1
```
