```python
using System;
using System.Collections.Generic;

class Program
{
    // Função 1
    static string verificar_estoque_critico(int quantidade)
    {
        if (quantidade < 5)
            return " [REPOSIÇÃO NECESSÁRIA]";
        return "";
    }

    // Função 2
    static void exibir_cabecalho()
    {
        Console.WriteLine("\n     SORVETERIA    ");
        Console.WriteLine("---------------------------");
    }

    static void Main()
    {
        List<string> nomesInsumos = new List<string>();
        List<int> quantidades = new List<int>();

        int i = 1;

        // while no cadastro
        while (i <= 3)
        {
            Console.Write($"Digite o nome do {i}º produto: ");
            string nome = Console.ReadLine();

            Console.Write($"Digite a quantidade de {nome}: ");
            int qtd = int.Parse(Console.ReadLine());

            nomesInsumos.Add(nome);
            quantidades.Add(qtd);

            i++;
        }

        exibir_cabecalho();

        // while no relatório
        int j = 0;
        while (j < nomesInsumos.Count)
        {
            string nome = nomesInsumos[j];
            int qtd = quantidades[j];

            string alerta = verificar_estoque_critico(qtd);

            Console.WriteLine($"Produto: {nome}{alerta} | Estoque: {qtd} unidades.");

            j++;
        }
    }
}
```
