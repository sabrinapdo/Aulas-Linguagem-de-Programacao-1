```python
import java.util.ArrayList;
import java.util.Scanner;

public class Desafio {

    // Função 1
    public static String verificar_estoque_critico(int quantidade) {
        if (quantidade < 5) {
            return " [REPOSIÇÃO NECESSÁRIA]";
        }
        return "";
    }

    // Função 2
    public static void exibir_cabecalho() {
        System.out.println("\n      SORVETERIA      ");
        System.out.println("---------------------------");
    }

    public static void main(String[] args) {
        Scanner scanner = new Scanner(System.in);

        ArrayList<String> nomesInsumos = new ArrayList<>();
        ArrayList<Integer> quantidades = new ArrayList<>();

        int i = 1;

        // while no cadastro
        while (i <= 3) {
            System.out.print("Digite o nome do " + i + "º produto: ");
            String nome = scanner.nextLine();

            System.out.print("Digite a quantidade de " + nome + ": ");
            int qtd = scanner.nextInt();
            scanner.nextLine();

            nomesInsumos.add(nome);
            quantidades.add(qtd);

            i++;
        }

        exibir_cabecalho();

        // while no relatório
        int j = 0;
        while (j < nomesInsumos.size()) {
            String nome = nomesInsumos.get(j);
            int qtd = quantidades.get(j);

            String alerta = verificar_estoque_critico(qtd);

            System.out.println("Produto: " + nome + alerta + " | Estoque: " + qtd + " unidades.");

            j++;
        }

        scanner.close();
    }
}
```
