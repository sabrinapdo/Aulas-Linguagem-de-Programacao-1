```python
#Exercício 1: Matriz Mágica de Múltiplos

A = [2, 4, 6, 8, 10, 12, 14, 16]

B = [0] * len(A)

for i in range(len(A)):
    B[i] = A[i] * 3

print("Vetor B:", B)
```

```python
#Exercício 2: Subtraindo Matrizes

A = []
B = []
C = []

print("Digite os 20 valores da matriz A:")
for i in range(20):
    valor = float(input(f"A[{i}]: "))
    A.append(valor)

print("\nDigite os 20 valores da matriz B:")
for i in range(20):
    valor = float(input(f"B[{i}]: "))
    B.append(valor)

for i in range(20):
    C.append(A[i] - B[i])

print("\nMatriz C (A - B):")
for i in range(20):
    print(f"C[{i}] = {C[i]}")
```
