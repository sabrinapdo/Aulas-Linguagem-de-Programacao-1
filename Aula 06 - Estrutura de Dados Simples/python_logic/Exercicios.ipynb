```python
#Exercício 1: Matriz Mágica de Múltiplos

A = [2, 4, 6, 8, 10, 12, 14, 16]

B = [0] * len(A)

for i in range(len(A)):
    B[i] = A[i] * 3

print("Vetor B:", B)
```

```python
#Exercício 2: Subtraindo Matrizes

A = []
B = []
C = []

print("Digite os 20 valores da matriz A:")
for i in range(20):
    valor = float(input(f"A[{i}]: "))
    A.append(valor)

print("\nDigite os 20 valores da matriz B:")
for i in range(20):
    valor = float(input(f"B[{i}]: "))
    B.append(valor)

for i in range(20):
    C.append(A[i] - B[i])

print("\nMatriz C (A - B):")
for i in range(20):
    print(f"C[{i}] = {C[i]}")
```

```python
#Exercício 3: O Poder dos Quadrados

A = []
B = []

for i in range(15):
    A.append(int(input("Digite: ")))

for i in range(15):
    B.append(A[i] ** 2)

print("A:", A)
print("B:", B)
```

```python
#4 - Desafio Final: O Painel de Controle do Dener (MVP)

nomes = []
quantidades = []

# cadastro
for i in range(3):
    nome = input("Digite o nome do produto: ")
    qtd = int(input("Digite a quantidade: "))

    nomes.append(nome)
    quantidades.append(qtd)

# relatório
print("\n--- RELATÓRIO ---")
for i in range(3):
    if quantidades[i] < 5:
        print(f"Produto: {nomes[i]} | Estoque: {quantidades[i]} unidades [REPOSIÇÃO NECESSÁRIA]")
    else:
        print(f"Produto: {nomes[i]} | Estoque: {quantidades[i]} unidades")
```
