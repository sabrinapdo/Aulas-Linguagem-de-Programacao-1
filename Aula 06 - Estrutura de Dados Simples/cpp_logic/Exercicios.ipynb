```python
#Exercício 1: Matriz Mágica de Múltiplos

#include <iostream>
using namespace std;

int main() {
    int A[8] = {2, 4, 6, 8, 10, 12, 14, 16};
    int B[8];

    // Preenchendo B
    for (int i = 0; i < 8; i++) {
        B[i] = A[i] * 3;
    }

    // Exibindo B
    cout << "Vetor B: ";
    for (int i = 0; i < 8; i++) {
        cout << B[i] << " ";
    }

    return 0;
}
```

```python
#Exercício 2: Subtraindo Matrizes

#include <iostream>
using namespace std;

int main() {
    double A[20], B[20], C[20];

    cout << "Digite os 20 valores da matriz A:\n";
    for(int i = 0; i < 20; i++) {
        cout << "A[" << i << "]: ";
        cin >> A[i];
    }

    cout << "\nDigite os 20 valores da matriz B:\n";
    for(int i = 0; i < 20; i++) {
        cout << "B[" << i << "]: ";
        cin >> B[i];
    }

    for(int i = 0; i < 20; i++) {
        C[i] = A[i] - B[i];
    }

    cout << "\nMatriz C (A - B):\n";
    for(int i = 0; i < 20; i++) {
        cout << "C[" << i << "] = " << C[i] << endl;
    }

    return 0;
}
```

```python
#Exercício 3: O Poder dos Quadrados

#include <iostream>
using namespace std;

int main() {
    int A[15], B[15];

    for (int i = 0; i < 15; i++) cin >> A[i];

    for (int i = 0; i < 15; i++) {
        B[i] = A[i] * A[i];
    }

    cout << "A: ";
    for (int i = 0; i < 15; i++) cout << A[i] << " ";

    cout << "\nB: ";
    for (int i = 0; i < 15; i++) cout << B[i] << " ";

    return 0;
}
```

```python
#4 - Desafio Final: O Painel de Controle do Dener (MVP)

using System;
using System.Collections.Generic;

class Program
{
    static void Main()
    {
        // Duas coleções: uma para nomes, outra para quantidades
        List<string> nomesInsumos = new List<string>();
        List<int> quantidades = new List<int>();

        // Laço para cadastrar 3 produtos
        for (int i = 1; i <= 3; i++)
        {
            Console.Write($"Digite o nome do {i}º produto: ");
            string nome = Console.ReadLine();

            Console.Write($"Digite a quantidade de {nome}: ");
            int qtd = int.Parse(Console.ReadLine());

            nomesInsumos.Add(nome);
            quantidades.Add(qtd);
        }

        Console.WriteLine("\n--- RELATÓRIO FINAL ---");

        // Exibe o relatório formatado
        for (int i = 0; i < nomesInsumos.Count; i++)
        {
            string nome = nomesInsumos[i];
            int qtd = quantidades[i];
            string alerta = (qtd < 5) ? " [REPOSIÇÃO NECESSÁRIA]" : "";

            Console.WriteLine($"Produto: {nome}{alerta} | Estoque: {qtd} unidades.");
        }
    }
}
```
