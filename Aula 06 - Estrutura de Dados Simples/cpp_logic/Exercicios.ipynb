```python
#Exercício 1: Matriz Mágica de Múltiplos

#include <iostream>
using namespace std;

int main() {
    int A[8] = {2, 4, 6, 8, 10, 12, 14, 16};
    int B[8];

    // Preenchendo B
    for (int i = 0; i < 8; i++) {
        B[i] = A[i] * 3;
    }

    // Exibindo B
    cout << "Vetor B: ";
    for (int i = 0; i < 8; i++) {
        cout << B[i] << " ";
    }

    return 0;
}
```

```python
#Exercício 2: Subtraindo Matrizes

#include <iostream>
using namespace std;

int main() {
    double A[20], B[20], C[20];

    cout << "Digite os 20 valores da matriz A:\n";
    for(int i = 0; i < 20; i++) {
        cout << "A[" << i << "]: ";
        cin >> A[i];
    }

    cout << "\nDigite os 20 valores da matriz B:\n";
    for(int i = 0; i < 20; i++) {
        cout << "B[" << i << "]: ";
        cin >> B[i];
    }

    for(int i = 0; i < 20; i++) {
        C[i] = A[i] - B[i];
    }

    cout << "\nMatriz C (A - B):\n";
    for(int i = 0; i < 20; i++) {
        cout << "C[" << i << "] = " << C[i] << endl;
    }

    return 0;
}
```
