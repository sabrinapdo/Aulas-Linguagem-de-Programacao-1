```python
#Exercício 1: Matriz Mágica de Múltiplos

public class Exercicios {
    public static void main(String[] args) {
        
        int[] A = {2, 4, 6, 8, 10, 12, 14, 16};
        int[] B = new int[8];

        // Preenchendo B
        for (int i = 0; i < A.length; i++) {
            B[i] = A[i] * 3;
        }

        // Exibindo B
        System.out.print("Vetor B: ");
        for (int i = 0; i < B.length; i++) {
            System.out.print(B[i] + " ");
        }
    }
}
```

```python
#Exercício 2: Subtraindo Matrizes

import java.util.Scanner;

public class Exercicios {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        double[] A = new double[20];
        double[] B = new double[20];
        double[] C = new double[20];

        System.out.println("Digite os 20 valores da matriz A:");
        for (int i = 0; i < 20; i++) {
            System.out.print("A[" + i + "]: ");
            A[i] = sc.nextDouble();
        }

        System.out.println("\nDigite os 20 valores da matriz B:");
        for (int i = 0; i < 20; i++) {
            System.out.print("B[" + i + "]: ");
            B[i] = sc.nextDouble();
        }

        for (int i = 0; i < 20; i++) {
            C[i] = A[i] - B[i];
        }

        System.out.println("\nMatriz C (A - B):");
        for (int i = 0; i < 20; i++) {
            System.out.println("C[" + i + "] = " + C[i]);
        }

        sc.close();
    }
}

```

```python
#Exercício 3: O Poder dos Quadrados

import java.util.Scanner;

public class Ex3 {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        int[] A = new int[15];
        int[] B = new int[15];

        for (int i = 0; i < 15; i++) {
            A[i] = sc.nextInt();
        }

        for (int i = 0; i < 15; i++) {
            B[i] = A[i] * A[i];
        }

        System.out.println("A:");
        for (int a : A) System.out.print(a + " ");

        System.out.println("\nB:");
        for (int b : B) System.out.print(b + " ");
    }
}
```

```python
#4 - Desafio Final: O Painel de Controle do Dener (MVP)

import java.util.ArrayList;
import java.util.Scanner;

public class ControleEstoque {
    public static void main(String[] args) {
        Scanner scanner = new Scanner(System.in);

        // Duas coleções: uma para nomes, outra para quantidades
        ArrayList<String> nomesInsumos = new ArrayList<>();
        ArrayList<Integer> quantidades = new ArrayList<>();

        // Laço para cadastrar 3 produtos
        for (int i = 1; i <= 3; i++) {
            System.out.print("Digite o nome do " + i + "º produto: ");
            String nome = scanner.nextLine();

            System.out.print("Digite a quantidade de " + nome + ": ");
            int qtd = scanner.nextInt();
            scanner.nextLine(); // limpar o buffer

            nomesInsumos.add(nome);
            quantidades.add(qtd);
        }

        System.out.println("\n--- RELATÓRIO FINAL ---");

        // Exibe o relatório formatado
        for (int i = 0; i < nomesInsumos.size(); i++) {
            String nome = nomesInsumos.get(i);
            int qtd = quantidades.get(i);
            String alerta = (qtd < 5) ? " [REPOSIÇÃO NECESSÁRIA]" : "";

            System.out.println("Produto: " + nome + alerta + " | Estoque: " + qtd + " unidades.");
        }

        scanner.close();
    }
}
```
