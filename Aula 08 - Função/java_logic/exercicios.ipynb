```python
#Exercicio 3 - Versaõ 2: Funções

import java.util.Scanner;

public class Main {

    static boolean validarSenha(String s){
        return s.length() >= 8;
    }

    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        String senha = "";

        while(!validarSenha(senha)){
            senha = sc.nextLine();
        }

        System.out.println("Senha válida!");
    }
}
```

```python
#Exercicio 6 - Versaõ 1: Procedural

import java.util.Scanner;

public class Main {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        double valor = sc.nextDouble();
        double desc = sc.nextDouble();

        double finalValor = valor - (valor * desc / 100);
        System.out.println("Final: " + finalValor);
    }
}
```

```python
#Exercicio 7 - Versaõ 1: Procedural

import java.util.Scanner;

public class Main {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        int nota = sc.nextInt();

        if(nota >= 90)
            System.out.println("A");
        else if(nota >= 80)
            System.out.println("B");
        else if(nota >= 70)
            System.out.println("C");
        else if(nota >= 60)
            System.out.println("D");
        else
            System.out.println("E");
    }
}
```

```python
#Exercicio 7 - Versaõ 2: Funções

static char classificar(int n){
    if(n >= 90) return 'A';
    else if(n >= 80) return 'B';
    else if(n >= 70) return 'C';
    else if(n >= 60) return 'D';
    else return 'E';
}
```

```python
#Exercicio 8 - Versaõ 1: Procedural

import java.util.Scanner;

public class Main {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        int a = sc.nextInt();
        int b = sc.nextInt();
        int c = sc.nextInt();

        if(a == b && b == c)
            System.out.println("Equilátero");
        else if(a == b || b == c || a == c)
            System.out.println("Isósceles");
        else
            System.out.println("Escaleno");
    }
}
```

```python
#Exercicio 8 - Versaõ 2: Funções

static String classificarTriangulo(int a, int b, int c){
    if(a == b && b == c)
        return "Equilátero";
    else if(a == b || b == c || a == c)
        return "Isósceles";
    else
        return "Escaleno";
}
```

```python
#Exercicio 9 - Versaõ 1: Procedural

import java.util.Scanner;

public class Main {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        String texto = sc.nextLine();
        int cont = 0;

        for(char c : texto.toLowerCase().toCharArray()){
            if("aeiou".indexOf(c) != -1)
                cont++;
        }

        System.out.println(cont);
    }
}
```

```python
#Exercicio 9 - Versão 2: Funções

static int contarVogais(String texto){
    int cont = 0;
    for(char c : texto.toLowerCase().toCharArray()){
        if("aeiou".indexOf(c) != -1)
            cont++;
    }
    return cont;
}
```
