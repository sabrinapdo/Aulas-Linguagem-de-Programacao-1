```python
#Exercicio 7 - Versaõ 1: Procedural

#include <iostream>
using namespace std;

int main(){
    int nota;
    cin >> nota;

    if(nota >= 90)
        cout << "A";
    else if(nota >= 80)
        cout << "B";
    else if(nota >= 70)
        cout << "C";
    else if(nota >= 60)
        cout << "D";
    else
        cout << "E";

    return 0;
}
```

```python
#Exercicio 7 - Versaõ 2: Funções

char classificar(int n){
    if(n >= 90) return 'A';
    else if(n >= 80) return 'B';
    else if(n >= 70) return 'C';
    else if(n >= 60) return 'D';
    else return 'E';
}
```

```python
#Exercicio 8 - Versaõ 1: Procedural

#include <iostream>
using namespace std;

int main(){
    int a, b, c;
    cin >> a >> b >> c;

    if(a == b && b == c)
        cout << "Equilatero";
    else if(a == b || b == c || a == c)
        cout << "Isosceles";
    else
        cout << "Escaleno";

    return 0;
}
```

```python
#Exercicio 8 - Versaõ 2: Funções

string classificar_triangulo(int a, int b, int c){
    if(a == b && b == c)
        return "Equilatero";
    else if(a == b || b == c || a == c)
        return "Isosceles";
    else
        return "Escaleno";
}
```

```python
#Exercicio 9 - Versaõ 1: Procedural

#include <iostream>
using namespace std;

int main(){
    string texto;
    getline(cin, texto);

    int cont = 0;

    for(char c : texto){
        c = tolower(c);
        if(c=='a'||c=='e'||c=='i'||c=='o'||c=='u')
            cont++;
    }

    cout << cont;

    return 0;
}
```

```python
#Exercicio 9 - Versão 2: Funções

int contar_vogais(string texto){
    int cont = 0;
    for(char c : texto){
        c = tolower(c);
        if(c=='a'||c=='e'||c=='i'||c=='o'||c=='u')
            cont++;
    }
    return cont;
}
```
