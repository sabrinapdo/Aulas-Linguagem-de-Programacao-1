```python
#Exercicio 1 - Versaõ 1: Procedural

num = int(input("Digite um número: "))

if num % 2 == 0:
    print("Par")
else:
    print("Ímpar")
```

```python
#Exercicio 1 - Versaõ 2: Funções

def verificar_paridade(n):
    if n % 2 == 0:
        print("Par")
    else:
        print("Ímpar")

num = int(input("Digite um número: "))
verificar_paridade(num)
```

```python
#Exercicio 2 - Versaõ 1: Procedural

base = float(input("Base: "))
altura = float(input("Altura: "))

area = base * altura
print("Área:", area)
```

```python
#Exercicio 2 - Versaõ 2: Funções

def calcular_area(base, altura):
    return base * altura

base = float(input("Base: "))
altura = float(input("Altura: "))

print("Área:", calcular_area(base, altura))
```

```python
#Exercicio 3 - Versaõ 1: Procedural

senha = ""

while len(senha) < 8:
    senha = input("Digite uma senha (mínimo 8 caracteres): ")

print("Senha válida!")
```

```python
#Exercicio 3 - Versaõ 2: Funções

def validar_senha(s):
    return len(s) >= 8

senha = ""

while not validar_senha(senha):
    senha = input("Digite uma senha (mínimo 8 caracteres): ")

print("Senha válida!")
```

```python
#Exercicio 5 - Versaõ 1: Procedural

c = float(input("Celsius: "))
f = (c * 9/5) + 32

print("Fahrenheit:", f)
```

```python
#Exercicio 5 - Versaõ 2: Funções

def converter(c):
    return (c * 9/5) + 32

c = float(input("Celsius: "))
print("Fahrenheit:", converter(c))
```

```python
#Exercicio 6 - Versaõ 1: Procedural

valor = float(input("Valor: "))
desc = float(input("Desconto (%): "))

final = valor - (valor * desc / 100)
print("Final:", final)
```

```python
#Exercicio 6 - Versaõ 2: Funções

def aplicar_desconto(valor, desc):
    return valor - (valor * desc / 100)

valor = float(input("Valor: "))
desc = float(input("Desconto (%): "))

print("Final:", aplicar_desconto(valor, desc))
```

```python
#Exercicio 7 - Versaõ 1: Procedural

nota = int(input())

if nota >= 90:
    print("A")
elif nota >= 80:
    print("B")
elif nota >= 70:
    print("C")
elif nota >= 60:
    print("D")
else:
    print("E")
```

```python
#Exercicio 7 - Versaõ 2: Funções

def classificar(n):
    if n >= 90: return 'A'
    elif n >= 80: return 'B'
    elif n >= 70: return 'C'
    elif n >= 60: return 'D'
    else: return 'E'

nota = int(input())
print(classificar(nota))
```

```python
#Exercicio 8 - Versaõ 1: Procedural

a = int(input())
b = int(input())
c = int(input())

if a == b == c:
    print("Equilátero")
elif a == b or b == c or a == c:
    print("Isósceles")
else:
    print("Escaleno")
```

```python
#Exercicio 8 - Versaõ 2: Funções

def triangulo(a,b,c):
    if a == b == c:
        return "Equilátero"
    elif a == b or b == c or a == c:
        return "Isósceles"
    else:
        return "Escaleno"

print(triangulo(a,b,c))
```

```python
#Exercicio 9 - Versaõ 1: Procedural

texto = input()
cont = 0

for c in texto.lower():
    if c in "aeiou":
        cont += 1

print(cont)
```

```python
#Exercicio 9 - Versaõ 2: Funções

def contar_vogais(t):
    cont = 0
    for c in t.lower():
        if c in "aeiou":
            cont += 1
    return cont

texto = input()
print(contar_vogais(texto))
```
